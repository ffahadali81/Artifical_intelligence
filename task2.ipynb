```python
numbers = [12, 7, 15, 20, 9, 8, 31, 14]

even = []
odd = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)

print("Original List:", numbers)
print("Even Numbers:", even)
print("Odd Numbers:", odd)
print("Total Even Numbers:", len(even))
print("Total Odd Numbers:", len(odd))
```
