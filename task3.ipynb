```python
secret = 7
attempts = 0

while True:
    guess = int(input("Guess the number (1-10): "))
    attempts = attempts + 1

    if guess == secret:
       print("Correct guess!")
       print("Attempts:", attempts)
       break

    elif guess < secret:
       print("Too low! Try again.")
    else:
       print("Too high! Try again.")
```
